# NumPy Basic

## NumPy (Numerical Python) is a Python library used for numerical computations.

```python
# mporting numpy

import numpy as np
```

### Functions to create NumPy arrays

#### 1 D array

```python
# creating numpy array
arr = np.array([1,2,3,4,5])
print(arr)

# checking type of this array
print(type(arr))
```

#### 2 D array

```python
a = np.array([[1,2,3,4],[5,6,7,8]])
print(a)
```

#### 3 D array

```python
b = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(b)
```

#### dtype

```python
# making array of float data type
d = np.array([1,2,3],dtype=float)
print(d)
```

```python
# making array of bool data type
e = np.array([1,2,3,4,5],dtype=bool)
print(e)
```

#### np.arange()

```python
# syntax -> np.arange(start,end,skip)

arr = np.arange(1,6,2)
print(arr)
```

#### reshape

```python
# reshape converts the shape of arrays
# syntax -> .rehsape(rows, cols)
arr = np.arange(1,11).reshape(5,2)
print(arr)
```

#### np.ones()

```python
arr = np.ones((2,3))
print(arr)
```

#### np.zeros()

```python
arr = np.zeros((3,4))
print(arr)
```

#### np.random()

```python
# .random() used to create arrays with random values

arr = np.random.random((3,4))
print(arr)
```

#### np.linspace

```python
# linspace generates ponts on equal distance within a given range 
# syntax -> np.linspace(lower range, upper range, no. of items that you want to generate)

arr = np.linspace(10,20,10)
print(arr)
```

#### np.identity

```python
# identity matrix -> who's diagonally items are 1

arr = np.identity(3)
print(arr)
```

### Array Attributes

```python
# creating arrays 

a1 = np.arange(10)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
print(a3)
```

#### ndim -> means no. of dimensions

```python
# syntax -> arrayName.ndim

# checking dimension of array a1
print(a1.ndim)

# a2
print(a2.ndim)

# a3
print(a3.ndim)
```

#### .shape

##### .shape is used to know size of the array in each dimension

```python
# a1
print(a1.shape)

# a2
print(a2.shape)

# a3
print(a3.shape)
```

### .size 

#### .size return the total number of elements in an array,

```python
# a1
print(a1.size)

# a2
print(a2.size)

# a3
print(a3.size)
```

### .itemsize

#### .itemsize tell how much size each item is occupying in the memory 

```python
# a1
print(a1.itemsize)

# a2
print(a2.itemsize)

# a3
print(a3.itemsize)
```

### Changing Datatypes

#### .astype

```python
# displaying original 
print(type(a3))

# converting into int32
a3.astype(np.int32)
```

### Array Operations

#### Scalar Operations

```python
# in scalar operations we operate with single scalar (value) on the numpy array

# Arithmetic operations

# multiplying with scalar

# multiplying array a1 with 2
print(a1 * 2)

# multiplying array a2 with 2
print(a2 * 2)

# multiplying array a3 with 2
print(a3 *2)
```

```python
# adding with scalar

# a1
print(a1 +5)

# a2
print(a2 +5)

# a3
print(a3 +5)
```

```python
# Relationsl Operations

# a2
a2 > 5 # now this will work item wise and check that each item in a2 is greater then 5 or not, if yes then print true otherwise false
```

```python
# now let's check if the elements in a1 is  == 4
a1 == 4
```

#### Vector Operations

```python
# when we do operations on two numpy arrays, this is called vector operations

# adding a1 and a2
a2 + a2
```

### Array Functions

```python
# creating random array 

b1 = np.random.random((3,3))
b1 = np.round(b1*100)  # .round -> number ko nearest (sabse paas wali) simple value pe le jana
b1
```

#### max

```python
# max gives the maximum value in the array
np.max(b1)
```

### min

```python
# min gives the minimum value in the array
np.min(b1)
```

### sum

```python
# sum function is used to find the sum of the entire array
np.sum(b1)
```

#### prod

```python
# prod (product) gives the product (multiplication) of all the elememnts present in the array
np.prod(b1)
```

```python
# now if we want to find the minumum / maximum of each row / columns -> we use axis

# in axis
# 0 -> cols
# 1 -> rows 


np.min(b1,axis=0)
```

### Statistical Operations

#### mean

```python
# mean of entire array
np.mean(b1)


# if we want to find the mean of cols / rows we can use axis
np.mean(b1,axis=0)
```

#### median

```python
# median gives us the center value of array
np.median(b1)
```

#### std

```python
# std (standard deviation)
np.std(b1)
```

#### var

```python
# variance
np.var(b1)
```

### Trigonometric functions

```python
# sin

np.sign(b1)

# similarly you can use cos, tan etc
```

### dot product

```python
# creating arrays

b2 = np.arange(12).reshape(3,4) # 3 rows and 4 cols
b3 = np.arange(12,24).reshape(4,3) # 4 rows and 3 cols

# basic rule for finding the dot product is (column of 1st should match the row of 2nd)

np.dot(b2,b3) # result will be in 3 X 3 matrix
```

#### logs and exponents

```python
# caluclating log
np.log(b2)
```

```python
# calculting ecponents using exp function
np.exp(b2)
```

#### round, floor and ceil

```python
# creating array

# round ka use numbers ko nearest value (usually integer ya specified decimal place) tak le jane ke liye hota hai. 

# in simple words -> it rounds off to the nearest integer value
np.round(np.random.random((2*3)) * 100)
```

```python
# floor
# means downward rounding (peeche ki trf le k jata h values ko)

np.floor(np.random.random((2,3))*100)
```

```python
# ceil
# ceil means upward rounding (aage wale interger p le k jata h values ko)
np.ceil(np.random.random((2,3))*100)
```
