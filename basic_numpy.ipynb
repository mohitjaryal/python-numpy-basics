# NumPy Basic

## NumPy (Numerical Python) is a Python library used for numerical computations.

```python
# mporting numpy

import numpy as np
```

### Functions to create NumPy arrays

#### 1 D array

```python
# creating numpy array
arr = np.array([1,2,3,4,5])
print(arr)

# checking type of this array
print(type(arr))
```

#### 2 D array

```python
a = np.array([[1,2,3,4],[5,6,7,8]])
print(a)
```

#### 3 D array

```python
b = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(b)
```

#### dtype

```python
# making array of float data type
d = np.array([1,2,3],dtype=float)
print(d)
```

```python
# making array of bool data type
e = np.array([1,2,3,4,5],dtype=bool)
print(e)
```

#### np.arange()

```python
# syntax -> np.arange(start,end,skip)

arr = np.arange(1,6,2)
print(arr)
```

#### reshape

```python
# reshape converts the shape of arrays
# syntax -> .rehsape(rows, cols)
arr = np.arange(1,11).reshape(5,2)
print(arr)
```

#### np.ones()

```python
arr = np.ones((2,3))
print(arr)
```

#### np.zeros()

```python
arr = np.zeros((3,4))
print(arr)
```

#### np.random()

```python
# .random() used to create arrays with random values

arr = np.random.random((3,4))
print(arr)
```

#### np.linspace

```python
# linspace generates ponts on equal distance within a given range 
# syntax -> np.linspace(lower range, upper range, no. of items that you want to generate)

arr = np.linspace(10,20,10)
print(arr)
```

#### np.identity

```python
# identity matrix -> who's diagonally items are 1

arr = np.identity(3)
print(arr)
```

### Array Attributes

```python
# creating arrays 

a1 = np.arange(10)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
print(a3)
```

```python
#### ndim -> means no. of dimensions
```

```python
# syntax -> arrayName.ndim

# checking dimension of array a1
print(a1.ndim)

# a2
print(a2.ndim)

# a3
print(a3.ndim)
```

#### .shape

##### .shape is used to know size of the array in each dimension

```python
# a1
print(a1.shape)

# a2
print(a2.shape)

# a3
print(a3.shape)
```

### .size 

#### .size return the total number of elements in an array,

```python
# a1
print(a1.size)

# a2
print(a2.size)

# a3
print(a3.size)
```

### .itemsize

#### .itemsize tell how much size each item is occupying in the memory 

```python
# a1
print(a1.itemsize)

# a2
print(a2.itemsize)

# a3
print(a3.itemsize)
```
